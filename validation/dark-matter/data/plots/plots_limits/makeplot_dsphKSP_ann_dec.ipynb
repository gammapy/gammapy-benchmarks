## 1/ Compare dSphs limits -- Annihilation Einasto and bb

```python
magic_22_mdsph_ann   = ascii.read(pathb + "magic_2022_multidsph_ann_bb.ecsv")
hawc_18_mdsph_ann    = ascii.read(pathb + "hawc_2018_multidsph_ann_bb_notri2.ecsv")
hess_20_mdsph_ann    = ascii.read(pathb + "hess_2020_multidsph_ann_bb.ecsv")
fermi_23_mdsph_ann   = ascii.read(pathb + "lat_2023_multidsph_ann_bb_measured.ecsv")
lhaaso_24_mdsph_ann  = ascii.read(pathb + "lhaaso_2024_multidsph_ann_bb.ecsv")
veritas_17_mdsph_ann = ascii.read(pathb +  "veritas_2017_multidsph_ann_bb.ecsv")


steigman = pd.read_csv('Relic_WIMP_Abundance.tsv',sep='\s+')
#steigman['mass'] =  steigman['mass'] / 1000

combined_einasto_600_ann = pd.read_csv(path_wsjc+'/combined_einasto_annihilation_DraI_UMi_t600_wsj_quantiles.csv')
```
